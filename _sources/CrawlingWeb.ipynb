```python
import requests
import pandas as pd

# API Key Springer Nature
api_key = "ae33356e10bc9a2e8384da9851a5f271"

# daftar kata kunci
keywords = ["web mining", "web usage mining"]

#alamat url
url = "https://api.springernature.com/meta/v2/json"

all_records = []

for kw in keywords:
    params = {
        "q": kw,
        "api_key": api_key,
        "p": 10
    }

    response = requests.get(url, params=params)

    if response.status_code == 200:
        data = response.json()
        print(f"Kata kunci: {kw} | Total hasil: {data['result'][0]['total']}\n")

        for record in data['records']:
            doi = record.get('doi', 'N/A')
            title = record.get('title', 'No title')
            abstract = record.get('abstract', 'No abstract')

            print(f"DOI: {doi}")
            print(f"Title: {title}")
            print(f"Abstract: {abstract}\n")

            all_records.append({
                "Keyword": kw,
                "DOI": doi,
                "Title": title,
                "Abstract": abstract
            })
    else:
        print("Error:", response.status_code, response.text)
```

```python
# simpan ke CSV gabungan
df = pd.DataFrame(all_records)
df.to_csv("Crawlingweb.csv", index=False, encoding="utf-8")
print("✅ Semua hasil sudah disimpan ke Crawlingweb.csv")
```
